```python
import pandas as pd

# Load data
df = pd.read_csv('../data/SpotifyFeatures.csv')

# Peek at the data
df.head()
```

```python
print("Shape:", df.shape)
df.info()
df.describe()
```

```python
# df.isnull().sum()
# df.duplicated().sum()
df['track_genre'].value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(24, 24))
sns.countplot(y='track_genre', data=df, order=df['track_genre'].value_counts().index)
plt.title("Number of Tracks per Genre")
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns

plt.figure(figsize=(12, 8))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm')
plt.title("Feature Correlation Matrix")
plt.show()
```

## Observations from Correlation Matrix

### Genre Representation
- Each genre has 1,000 tracks.
- No genre is over- or underrepresented.

### Strong Correlations
- `energy` and `loudness`: 0.76  
  Louder songs tend to be more energetic.
- `acousticness` and `energy`: -0.73  
  Acoustic songs tend to have lower energy.
- `danceability` and `valence`: 0.48  
  Songs that are more danceable also tend to be more positive in tone.
- `instrumentalness` and `loudness`: -0.43  
  Instrumental songs tend to be quieter.

## Other Notes
- `Unnamed: 0` does not relate meaningfully to any other column. It appears to be an index and can be removed.
- `mode` and `key` show very low correlation with all other features. This suggests they carry little information that overlaps with the rest of the dataset. Whether they are useful or not will depend on the modeling task, but they do not interact strongly with other variables.
- Most correlation values fall between -0.2 and 0.2, indicating that many features are weakly related. This means the dataset contains mostly independent variables, which reduces the risk of overlap but may also limit the ability to find strong predictive patterns without more complex modeling.
